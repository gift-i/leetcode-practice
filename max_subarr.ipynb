## Maximum Subarray

Given an integer array [nums], find the subarray with the largest sum, and return its sum.

```python
class Solution:
    def __init__ (self, nums): # do not need to be here for leetcode
        self.nums = nums # do not need to be here for leetcode
    def max_subarr (self): # replace with     def maxSubArray(self, nums: List[int]) -> int:
        sums = []
        maximum = self.nums.index(max(self.nums))
        sums.append(self.nums[maximum])
        for i in range(0, len(self.nums)):
            for j in range(0, len(self.nums)):
                upper_index = maximum + i
                lower_index = maximum - j
                if upper_index > len(self.nums)-1:
                    upper_index == len(self.nums)-1
                if lower_index < 0:
                    lower_index == 0
                sums.append(sum(self.nums[lower_index:upper_index]))
        return (max(sums))
```

```python
a = Solution(nums = [1])
print(a.max_subarr())
```
