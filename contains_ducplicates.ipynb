## Contains Duplicates

Given an integer array [nums], return [true] if any value appears at least twice in the array, and return [false] if every element is distinct.

```python
class ContainsDuplicates:
    def __init__(self, nums):
        self.nums = nums
    def solution(self):
        count = 0
        for i in self.nums:
            if self.nums.count(i)>1:
                count +=1
        if count>0:
            ans = True
        else:
            ans = False
        return ans
```

```python
a = ContainsDuplicates(nums = [1,2,3,0,36,4,54,6])
print(a.solution())
```
