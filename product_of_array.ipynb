## Product of Array Except Self

Given an integer array nums, return an array answer such that answer[i] is equal to the product of all the elements of nums except nums[i].

The product of any prefix or suffix of nums is guaranteed to fit in a 32-bit integer.

You must write an algorithm that runs in O(n) time and without using the division operation.

```python
class Solution:
    def __init__ (self, nums):
        self.nums = nums
    def product_of_array(self):
        new_nums = [] # to create the new list for the items
        import math
        for i in self.nums:
            index = self.nums.index(i)
            self.nums.remove(i) # remove the element in the list
            new_nums.append(math.prod(self.nums))
            self.nums.insert(index, i) # put back the item in the list (index, item)
        return new_nums
```

```python
a = Solution(nums = [1,2,3,4])
print(a.product_of_array())
```
