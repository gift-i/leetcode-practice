## Sum of Two Integers

Given two integers a and b, return the sum of the two integers without using the operators + and -.

```python
class Solution:
    def __init__ (self, a, b):
        self.a = a
        self.b = b
    def getSum(self):
        import math
        return int(math.log(math.e**(self.a) * math.e**(self.b)))
```

```python
print(Solution(1,2).getSum())
```
