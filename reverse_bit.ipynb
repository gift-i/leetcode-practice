## Reverse Bits

Reverse bits of a given 32 bits signed integer.

```python
class Solution:
    def __init__(self, n):
        self.n = n
    def reverse_bit (self):
        binary = bin(self.n).replace('0b','')
        while len(binary)<32:
            binary = f'0{binary}'
        reverse = str(binary)[::-1] # reverse the order of the string
        number = int(reverse, 2)
        return number
```

```python
print(Solution(43261596).reverse_bit())
```
