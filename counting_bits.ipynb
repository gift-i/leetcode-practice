## Counting Bits

Given an integer n, return an array [ans] of length [n + 1] such that for each i (0 <= i <= n), ans[i] is the number of 1's in the binary representation of i.

```python
class Solution:
    def __init__ (self, n):
        self.n = n
    def countBits(self):
        ans = []
        for i in range(0,self.n + 1): # +1 to include the actual number
            binary = bin(i)
            count = 0
            for j in binary:
                if j == '1':
                    count += 1
            ans.append(count)
        return ans
```

```python
print(Solution(5).countBits())
```
